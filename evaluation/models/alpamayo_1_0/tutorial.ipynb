## Docker steps
1. Build

```shell
cd evaluation/models/alpamayo_1_0/
docker compose -f docker-compose.yml --env-file /path/to/your/env_files/models/alpamayo_1_0/dev.env build
```

2. Start with jupyter service

```shell
docker compose -f docker-compose.yml --env-file /path/to/your/env_files/models/alpamayo_1_0/dev.env --profile jupyter up -d jupyter

```

3. Print the docker logs to get the jupyter server link

```shell
docker logs <container-name-or-id> 
docker logs alpamayo_1_0-jupyter-1 
```
